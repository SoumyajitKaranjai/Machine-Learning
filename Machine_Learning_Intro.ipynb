Question 1: Explain the differences between AI, ML, Deep Learning (DL), and Data Science (DS).

**Artificial Intelligence (AI):**

a. AI is the broadest concept that refers to machines or systems that imitate human intelligence and behavior.

b. It involves creating intelligent machines capable of learning, reasoning, problem-solving, perception, language understanding, and decision-making.

c. AI is a field of computer science that includes rule-based systems, knowledge-based systems, and data-driven approaches.

Example applications: AI-powered predictions, speech recognition, self-driving cars.

**Machine Learning (ML):**

a. ML is a subset of AI focused on designing algorithms that allow machines to learn from data and improve over time without explicit programming.

b. It involves training models on data to identify patterns and make predictions or decisions.

c. ML uses statistical methods and is commonly categorized into supervised, unsupervised, and reinforcement learning.

Example applications: Spam filtering, virtual personal assistants like Siri and Alexa.

**Deep Learning (DL):**

a. DL is a specialized subset of ML that uses deep neural networks with multiple layers to model complex patterns in large datasets.

b. It mimics the human brain's structure and functioning and is capable of automatic feature extraction.

c. DL generally requires large amounts of data and significant computational resources.

Example applications: Image recognition, speech recognition, cancer tumor detection.

**Data Science (DS):**

a. DS is an interdisciplinary field that uses scientific methods, processes, algorithms, and systems to extract knowledge and insights from structured and unstructured data.

b. It combines statistics, data analysis, machine learning, and domain knowledge to analyze and interpret data for decision-making and predictions.

c. DS often employs AI and ML techniques as part of its toolset but has a broader focus on data processing and analysis.

Example tasks: Data cleaning, exploratory data analysis (EDA), visualization, predictive analytics.

Question 2: What are the types of machine learning? Describe each with one
real-world example.

The main types of machine learning are:

**Supervised Learning:**

a. The model is trained on labeled data, meaning the input data comes with corresponding correct output labels.

b. The goal is to learn a mapping from inputs to outputs to make predictions on new data.

c. Includes two main tasks: classification (predicting discrete labels) and regression (predicting continuous values).

Example: Email spam detection. The model learns from emails labeled as "spam" or "not spam" and then classifies new incoming emails accordingly.

**Unsupervised Learning:**

a. The model is trained on unlabeled data without explicit output labels.

b. The goal is to find patterns, groupings, or structures in the data.

c. Common tasks are clustering and dimensionality reduction.

Example: Customer segmentation. The model groups customers with similar buying behaviors without prior knowledge of group labels.

**Reinforcement Learning:**

a. The model learns by interacting with an environment and receiving feedback in the form of rewards or penalties.

b. The goal is to learn a policy that maximizes cumulative future rewards.

c. Used in decision-making problems with sequential steps.

Example: Training a robot to navigate a maze, where it learns from reward signals when reaching closer to the exit.

Question 3: Define overfitting, underfitting, and the bias-variance tradeoff in machine learning.

**Overfitting:**

a. Overfitting occurs when a model learns the training data too well, capturing not only the underlying patterns but also noise and irrelevant details.

b. As a result, it performs extremely well on the training data but poorly on new, unseen data because it cannot generalize.

c. Causes include overly complex models, too little training data, noisy data, or training for too many iterations.

Example: A model trained to identify dogs may memorize background features like grass from the training photos and fail to recognize dogs in different settings.

**Underfitting:**

a. Underfitting happens when a model is too simple to learn the underlying pattern of the data.

b. It results in poor performance on both the training data and new data.

c. Causes include models that are too simple, insufficient training, or lack of relevant features.

Example: Using a linear model to fit data with a complex nonlinear relationship may fail to capture the trend, leading to inaccurate predictions.

**Bias-Variance Tradeoff:**

a. This is a fundamental concept describing the tradeoff between a model's ability to minimize bias and variance.

b. Bias refers to error due to overly simplistic assumptions in the model (leading to underfitting).

c. Variance refers to error due to the model being too sensitive to small fluctuations in the training data (leading to overfitting).

d. A good model balances bias and variance to achieve low total error and generalize well to new data.

Question 4: What are outliers in a dataset, and list three common techniques for handling them.

**Outliers:-**

Outliers in a dataset are data points that differ significantly from most other observations. They may represent rare or abnormal values, measurement errors, or novel insights. Outliers can distort statistical analyses and affect model performance if not properly handled.

Three common techniques for handling outliers are:

**Removal:**

a. Simply exclude outliers from the dataset when they are identified as errors or irrelevant anomalies.

b.Useful when outliers are due to data entry or measurement mistakes.

**Transformation:**

a. Apply mathematical transformations (e.g., log, square root) to reduce the impact of outliers.

b. This can make the distribution more normal and lessen the influence of extreme values.

**Imputation:**

a. Replace outliers with a calculated value like the mean, median, or a value from a predictive model.

b. This can help preserve data size while reducing the distortion caused by outliers.

Question 5: Explain the process of handling missing values and mention one
imputation technique for numerical and one for categorical data.

Handling missing values in a dataset involves several steps:

a.Identifying missing values - recognizing which data points are missing or null.

b. Analyzing the pattern and cause of missingness to decide on the appropriate handling strategy.

c. Choosing a strategy for managing missing values based on the data type and the extent of missingness.

d. Applying the chosen technique to impute or fill missing values to maintain data integrity and improve model performance.

One common imputation technique for numerical data is:

*    Mean Imputation: Replace missing numerical values with the mean of the available values in that feature. This method is simple and quick, suitable when data are missing completely at random and there are no extreme outliers.

One common imputation technique for categorical data is:

*  Mode Imputation: Replace missing categorical values with the most
frequent category (mode) in that feature. This maintains category distribution and works well with nominal data.

Question 8: Write a Python script to:

● Generate 1000 samples from a normal distribution.

● Introduce 50 random missing values.

● Fill missing values with the column mean.

● Plot a histogram before and after imputation.



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

#  Generate 1000 samples from a normal distribution
data = np.random.normal(loc=0, scale=1, size=1000)
df = pd.DataFrame(data, columns=['Value'])

#  Introduce 50 random missing values
np.random.seed(42)
missing_indices = np.random.choice(df.index, size=50, replace=False)
df.loc[missing_indices, 'Value'] = np.nan

# Plot histogram before imputation
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.hist(df['Value'].dropna(), bins=30, color='skyblue', edgecolor='black')
plt.title('Histogram before Imputation')
plt.xlabel('Value')
plt.ylabel('Frequency')

#  Fill missing values with the column mean
mean_value = df['Value'].mean()
df['Value'].fillna(mean_value, inplace=True)

# Plot histogram after imputation
plt.subplot(1, 2, 2)
plt.hist(df['Value'], bins=30, color='lightgreen', edgecolor='black')
plt.title('Histogram after Imputation')
plt.xlabel('Value')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

Question 10: You are working as a data scientist for a retail company. You receive a customer transaction dataset that contains:

● Missing ages,

● Outliers in transaction amount,

● A highly imbalanced target (fraud vs. non-fraud),

● Categorical variables like payment method.
Explain the step-by-step data preparation plan you’d follow before training a machine learning model. Include how you’d address missing data, outliers, imbalance, and encoding.

Step 1: Handle Missing Values (Ages)
Analyze missing data pattern.

Impute missing ages, e.g., with median or mean (median is often robust to outliers).

Step 2: Handle Outliers (Transaction Amount)
Detect outliers using methods like IQR or Z-score.

Treat outliers by capping (winsorizing) or removing based on business context.

Step 3: Handle Class Imbalance (Fraud vs. Non-Fraud)
Use techniques like SMOTE (Synthetic Minority Over-sampling Technique) or class-weight adjustment to balance classes.

Step 4: Encode Categorical Variables (Payment Method)
Use one-hot encoding to convert categorical variables into numeric format suitable for ML algorithms.

Step 5: Scale Numerical Features
Scale features like transaction amount and age using MinMaxScaler or StandardScaler.

```python
!pip install imblearn
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE
from scipy import stats

# Assume df is your loaded DataFrame
# Example structure for demonstration
df = pd.DataFrame({
    'Age': [25, np.nan, 37, 29, 45, np.nan, 30],
    'TransactionAmount': [100, 1200, 150, 200, 10000, 180, 170],  # 10000 might be an outlier
    'PaymentMethod': ['Credit Card', 'Debit Card', 'Debit Card', 'Paypal', 'Paypal', 'Credit Card', 'Credit Card'],
    'Fraud': [0, 0, 0, 1, 1, 0, 0]
})

# Step 1: Impute missing Ages with median
age_imputer = SimpleImputer(strategy='median')
df['Age'] = age_imputer.fit_transform(df[['Age']])

# Step 2: Detect and cap outliers in TransactionAmount using IQR
Q1 = df['TransactionAmount'].quantile(0.25)
Q3 = df['TransactionAmount'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
df['TransactionAmount'] = np.where(df['TransactionAmount'] > upper_bound, upper_bound,
                                   np.where(df['TransactionAmount'] < lower_bound, lower_bound, df['TransactionAmount']))

# Step 3: Separate features and target for oversampling
X = df.drop('Fraud', axis=1)
y = df['Fraud']

# Preprocessing pipeline for numeric and categorical features
numeric_features = ['Age', 'TransactionAmount']
categorical_features = ['PaymentMethod']

# One-hot encode categorical features and scale numeric features
preprocessor = ColumnTransformer(
    transformers=[
        ('num', MinMaxScaler(), numeric_features),
        ('cat', OneHotEncoder(drop='first'), categorical_features)
    ])

# Apply preprocessing
X_processed = preprocessor.fit_transform(X)

# Step 4: Address imbalance using SMOTE
smote = SMOTE(random_state=42)
X_resampled, y_resampled = smote.fit_resample(X_processed, y)

# Final shapes after resampling
print("Before resampling class distribution:\n", y.value_counts())
print("After resampling class distribution:\n", pd.Series(y_resampled).value_counts())
print("Processed feature array shape:", X_resampled.shape)
```
